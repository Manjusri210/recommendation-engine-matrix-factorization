```python
from surprise import Dataset

data = Dataset.load_builtin('ml-100k', prompt=False)
```

```python
from surprise import SVD
from surprise.model_selection import train_test_split

trainset, testset = train_test_split(data, test_size=0.2)

model = SVD()
model.fit(trainset)
```

```python
from surprise import accuracy

predictions = model.test(testset)
accuracy.rmse(predictions)
```

```python
def recommend_movies(user_id, n=5):
    all_items = set(trainset.all_items())
    rated_items = set([j for (j, _) in trainset.ur[user_id]])

    predictions = []

    for item in all_items - rated_items:
        pred = model.predict(user_id, item)
        predictions.append((item, pred.est))

    predictions.sort(key=lambda x: x[1], reverse=True)

    return predictions[:n]

print("Top Recommendations for User 1:")
print(recommend_movies(1))
```

Project Title: Recommendation Engine using Matrix Factorization

Dataset: MovieLens (ml-100k)

Description:
This project builds a recommendation system using Matrix Factorization (SVD).
It predicts user ratings and recommends movies based on user preferences.

Evaluation:
